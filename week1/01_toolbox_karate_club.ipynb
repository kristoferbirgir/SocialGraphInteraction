# Week 1 · Toolbox shakedown — Zachary's Karate Club

Exercise 1.3 ("Toolbox shakedown"). Load NetworkX's built-in Karate Club graph, draw it, and report its node count, edge count, and five highest-degree nodes.

```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.karate_club_graph()
print(f"nodes: {G.number_of_nodes()}")
print(f"edges: {G.number_of_edges()}")
```

## Draw the network

```python
pos = nx.spring_layout(G, seed=42)
plt.figure(figsize=(7, 7))
nx.draw_networkx(G, pos, node_color="#4C72B0", edge_color="#999999", node_size=350, font_size=8)
plt.title("Zachary's Karate Club")
plt.axis("off")
plt.show()
```

## Five highest-degree nodes

```python
degrees = dict(G.degree())
top5 = sorted(degrees.items(), key=lambda item: item[1], reverse=True)[:5]
for node, degree in top5:
    print(f"node {node}: degree {degree}")
```
